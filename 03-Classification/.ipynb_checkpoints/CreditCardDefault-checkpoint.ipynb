```python
import numpy as np
import pandas as pd

print(f"Numpy: {np.__version__}")
print(f"Pandas: {pd.__version__}")
```

```python
path = "default_data.csv"

df = pd.read_csv(path)
df.head()
```

```python
df.describe()
```

```python
# First, we have an issue where the columns are labelled "Xi" instead of useful names
list(df.iloc[0])
```

```python
df.columns = list(df.iloc[0])
```

```python
df.head()
```

```python
df = df[1:]
```

```python
df = df.rename(columns={'default payment next month': 'DEFAULT'})
df.head()
```

```python
del df["ID"]
```

```python
df.head()
```

```python
for col in df.columns:
    df[col] = df[col].astype(int)
```

```python
df.head()
```

**Data has now been formatted as integers, the column names have been set to appropriate names, and the ID field has been removed. The data is now ready to be put into a model**

```python
# The next step is to split the data into training, validation, and test sets
from sklearn.model_selection import train_test_split
```

```python
train_test_split?
```

```python
# Use a 60, 20, 20 split
val_n = 0.2
random_state=10

df_full_train, df_test = train_test_split(df, test_size=val_n, random_state=random_state)
```

```python
df_full_train.shape, df_test.shape
```

```python
df_train, df_val = train_test_split(df_full_train, test_size=(val_n)/(1-val_n), random_state=random_state)
```

```python
df_train.shape[0], df_val.shape[0], df_test.shape[0]
```

```python
y_full_train = df_full_train.DEFAULT.values
y_train = df_train.DEFAULT.values
y_val = df_val.DEFAULT.values
y_test = df_test.DEFAULT.values
```

```python
del df_full_train["DEFAULT"]
del df_train["DEFAULT"]
del df_val["DEFAULT"]
del df_test["DEFAULT"]
```

```python
df_full_train.head()
```

```python
y_full_train[0:10]
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=2000, solver='liblinear')
```

```python
model.fit(df_train, y_train)
```

```python
# Coefficients from a solver that did converge:
model.coef_
```

```python
solvers = ["lbfgs", "liblinear", "newton-cg", "newton-cholesky", "sag", "saga"]
```

```python
accuracies = []
for s in solvers:
    model = LogisticRegression(max_iter=2000, solver=s)
    model.fit(df_train, y_train)
    y_pred = model.predict_proba(df_val)[:, 1]

    accuracy = (y_val == (y_pred >= 0.5)).mean()
    accuracies.append(accuracy)
```

```python
for i, s in enumerate(solvers):
    print(f"{s}: {round(accuracies[i] * 100, 2)}")
```
