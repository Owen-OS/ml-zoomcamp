```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/alexeygrigorev/mlbookcamp-code/master/chapter-02-car-price/data.csv')
df.head()
```

```python
cols = [
    'Make',
    'Model',
    'Year',
    'Engine HP',
    'Engine Cylinders',
    'Transmission Type',
    'Vehicle Style',
    'highway MPG',
    'city mpg',
    'MSRP'
]
```

```python
df = df[cols]
df.head()
```

```python
df.columns = df.columns.str.replace(' ', '_').str.lower()
df.head()
```

```python
df = df.fillna(0)
```

```python
df.rename(columns={'msrp': 'price'}, inplace=True)
```

```python
df.head()
```

```python
df["above_average"] = (df["price"] > df["price"].mean()).astype(int)
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
n = 0.2
```

```python
df_full_train, df_test = train_test_split(df, test_size=n, random_state=42)
df_train, df_val = train_test_split(df_full_train, test_size=n/(1-n), random_state=42)
```

```python
df_train.shape, df_val.shape, df_test.shape
```

```python
y_full_train = df_full_train.above_average.values
y_train = df_train.above_average.values 
y_val = df_val.above_average.values 
y_test = df_test.above_average.values
```

```python
del df_train['price']
del df_val['price']
del df_test['price']
del df_full_train['price']

del df_train['above_average']
del df_val['above_average']
del df_test['above_average']
del df_full_train['above_average']
```

```python
from sklearn.metrics import mutual_info_score
```

```python
cat_cols = df_full_train.select_dtypes(include=['object']).columns.tolist()
cat_cols
```

```python
num_cols = df_full_train.select_dtypes(include=['number']).columns.tolist()
num_cols
```

```python
mutual_info_score?
```

```python
for col in cat_cols:
    mi = mutual_info_score(df_full_train[col], y_full_train)
    print(f'{col}: {mi:.2f}')
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction import DictVectorizer
```

```python
dv = DictVectorizer(sparse=False)
```

```python
train_dicts = df_train[cat_cols + num_cols].to_dict(orient='records')
val_dicts = df_val[cat_cols + num_cols].to_dict(orient='records')
test_dicts = df_test[cat_cols + num_cols].to_dict(orient='records')
```

```python
dv.fit(train_dicts)
```

```python
dv.get_feature_names_out()
```

```python
X_train = dv.transform(train_dicts)
X_val = dv.transform(val_dicts)
X_test = dv.transform(test_dicts)
```

```python
X_train[:5]
```

```python
model = LogisticRegression(solver='liblinear', C=10, max_iter=1000, random_state=42)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict_proba(X_val)[:, 1]
```

```python
round((y_val == (y_pred >= 0.5).astype(int)).mean(), 2)
```
