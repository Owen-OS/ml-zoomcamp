```python
num_cols = df.select_dtypes("number").columns.to_list()
len(num_cols)
```

```python
corr_matrix = df[num_cols].corr()
corr_matrix
```

```python
max_corr = 0
max_pair = None

for i, col in enumerate(num_cols):
    for j in range(i):
        if abs(corr_matrix.iloc[i, j]) > max_corr:
            max_corr = abs(corr_matrix.iloc[i, j])
            max_pair = (corr_matrix.index[i], corr_matrix.columns[j])   
            #print(f"{corr_matrix.index[i]} and {corr_matrix.columns[j]}: {corr_matrix.iloc[i, j]}") 
    
print(f"The most correlated columns are `{max_pair[0]}` and `{max_pair[1]}` with a correlation of {max_corr}")
```
